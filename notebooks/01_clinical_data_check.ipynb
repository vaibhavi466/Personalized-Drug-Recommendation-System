```python
import pandas as pd
from pathlib import Path

clinical_path = Path("../data/raw/gbm_tcga_pan_can_atlas_2018_clinical.tsv")

clinical = pd.read_csv(clinical_path, sep="\t")

print("Rows:", clinical.shape[0])
print("Columns:", clinical.shape[1])
```

```python
clinical[["Patient ID", "Sample ID", "Sample Type"]].head(10)
```

```python
samples_per_patient = clinical.groupby("Patient ID")["Sample ID"].nunique()

samples_per_patient.value_counts().sort_index()

# 578 patients have 1 sample only, 7 patients have 2 samples 
```

```python
clinical[clinical["Patient ID"].isin(
    samples_per_patient[samples_per_patient > 1].index
)][["Patient ID", "Sample ID", "Sample Type"]].sort_values("Patient ID")

# Actual 7 patients with multiple (2) samples == 14 records."
```

```python
clinical["Sample Type"].value_counts(dropna=False)
```

```python
primary_clinical = clinical[
    clinical["Sample Type"] == "Primary"
].copy()

print("Primary samples:", primary_clinical["Sample ID"].nunique())
print("Primary patients:", primary_clinical["Patient ID"].nunique())
print("Rows:", primary_clinical.shape[0])

# primary_clinical stores only primary samples i.e. 578 samples from 578 patients, no duplicates
```

```python
primary_clinical["Patient ID"].duplicated().sum()
```

```python
from pathlib import Path

processed_path = Path("../data/processed/gbm_primary_clinical.tsv")

primary_clinical.to_csv(
    processed_path,
    sep="\t",
    index=False
)

print("Saved to:", processed_path)
print("Exists:", processed_path.exists())

# Save the processed file to the data/processed directory for future use.
```
