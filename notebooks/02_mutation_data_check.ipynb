```python
import pandas as pd
from pathlib import Path

mutation_path = Path("../data/raw/gbm_tcga_pan_can_atlas_2018_mutations.txt")

print("Exists:", mutation_path.exists())
```

```python
mutations = pd.read_csv(
    mutation_path,
    sep="\t",
    comment="#",
    low_memory=False
)

print("Rows:", mutations.shape[0])
print("Columns:", mutations.shape[1])
```

```python
mutations["Mutation_Status"].value_counts(dropna=False)
```

```python
clinical_path = Path("../data/processed/gbm_primary_clinical.tsv")

primary_clinical = pd.read_csv(
    clinical_path,
    sep="\t"
)

print("Primary clinical samples:",
      primary_clinical["Sample ID"].nunique())

print("Mutation samples:",
      mutations["Tumor_Sample_Barcode"].nunique())
```

```python
primary_sample_ids = set(primary_clinical["Sample ID"])
mutation_sample_ids = set(mutations["Tumor_Sample_Barcode"])

overlap = primary_sample_ids & mutation_sample_ids

print("Samples present in both:", len(overlap))
```

```python
mutation_not_in_primary = mutation_sample_ids - primary_sample_ids

print("Mutation samples not in primary cohort:",
      len(mutation_not_in_primary))

mutation_not_in_primary

# 7 non-primary samples are exactly the same recurrent samples we identified earlier
```

```python
primary_mutations = mutations[
    mutations["Tumor_Sample_Barcode"].isin(primary_sample_ids)
].copy()

print("Mutation rows before filtering:", mutations.shape[0])
print("Mutation rows after primary filtering:", primary_mutations.shape[0])

print(
    "Unique primary mutation samples:",
    primary_mutations["Tumor_Sample_Barcode"].nunique()
)
```

```python
primary_mutations[
    primary_mutations["Tumor_Sample_Barcode"].str.endswith("-02")
]["Tumor_Sample_Barcode"].unique()
```

```python
primary_mutations["Variant_Classification"].value_counts()
```

```python
primary_mutations["Variant_Type"].value_counts()
```

```python
primary_mutations[
    ["Hugo_Symbol", "Variant_Classification", "Variant_Type"]
].drop_duplicates().head(20)
```

```python
protein_altering_classes = [
    "Missense_Mutation",
    "Nonsense_Mutation",
    "Frame_Shift_Del",
    "Frame_Shift_Ins",
    "Splice_Site",
    "In_Frame_Del",
    "In_Frame_Ins",
    "Translation_Start_Site",
    "Nonstop_Mutation"
]

protein_altering_mutations = primary_mutations[
    primary_mutations["Variant_Classification"].isin(protein_altering_classes)
].copy()

print("Primary mutation rows before functional filtering:",
      primary_mutations.shape[0])

print("Protein-altering mutation rows:",
      protein_altering_mutations.shape[0])

print("Unique samples:",
      protein_altering_mutations["Tumor_Sample_Barcode"].nunique())

print("Unique genes:",
      protein_altering_mutations["Hugo_Symbol"].nunique())
```

```python
protein_altering_mutations[
    "Variant_Classification"
].value_counts()
```

```python
protein_altering_mutations[
    ["Hugo_Symbol", "Variant_Classification", "Tumor_Sample_Barcode"]
].head(10)
```

```python
all_primary_mutation_samples = set(
    primary_mutations["Tumor_Sample_Barcode"]
)

protein_altering_samples = set(
    protein_altering_mutations["Tumor_Sample_Barcode"]
)

samples_without_protein_altering = (
    all_primary_mutation_samples - protein_altering_samples
)

print(
    "Samples with mutation records but no selected protein-altering mutations:",
    len(samples_without_protein_altering)
)

samples_without_protein_altering
```

```python
primary_mutations[
    primary_mutations["Tumor_Sample_Barcode"].isin(
        samples_without_protein_altering
    )
][
    [
        "Tumor_Sample_Barcode",
        "Hugo_Symbol",
        "Variant_Classification"
    ]
].sort_values(
    ["Tumor_Sample_Barcode", "Variant_Classification"]
)
```

```python
mutation_matrix = (
    protein_altering_mutations
    .assign(mutated=1)
    .pivot_table(
        index="Tumor_Sample_Barcode",
        columns="Hugo_Symbol",
        values="mutated",
        aggfunc="max",
        fill_value=0
    )
)

print("Rows (samples):", mutation_matrix.shape[0])
print("Columns (genes):", mutation_matrix.shape[1])
```

```python
mutation_matrix.head()
```

```python
all_primary_samples = primary_clinical["Sample ID"].tolist()

mutation_matrix_all = mutation_matrix.reindex(
    all_primary_samples,
    fill_value=0
)

print("Rows after adding all primary samples:",
      mutation_matrix_all.shape[0])

print("Columns:",
      mutation_matrix_all.shape[1])

# Add back all primary patients, even if they have no protein-altering mutations. 
```

```python
mutation_matrix_all.loc[
    ["TCGA-06-0165-01", "TCGA-06-0167-01"]
].sum(axis=1)
```

```python
zero_mutation_samples = (
    mutation_matrix_all.sum(axis=1) == 0
)

print("Samples with zero selected protein-altering mutations:",
      zero_mutation_samples.sum())
```

```python
from pathlib import Path

sequenced_path = Path(
    "../data/raw/gbm_tcga_pan_can_atlas_2018_cases_sequenced.txt"
)

print("Exists:", sequenced_path.exists())
```

```python
lines = sequenced_path.read_text().splitlines()

for line in lines[:6]:
    print(line[:300])
```

```python
case_ids_line = next(
    line for line in lines
    if line.startswith("case_list_ids:")
)

sequenced_sample_ids = set(
    case_ids_line.split(":", 1)[1].strip().split("\t")
)

print("Official mutation-profiled samples:",
      len(sequenced_sample_ids))
```

```python
primary_sample_ids = set(primary_clinical["Sample ID"])

sequenced_primary_ids = (
    sequenced_sample_ids & primary_sample_ids
)

sequenced_not_primary = (
    sequenced_sample_ids - primary_sample_ids
)

print("Sequenced primary samples:",
      len(sequenced_primary_ids))

print("Sequenced samples not in primary cohort:",
      len(sequenced_not_primary))

sequenced_not_primary
```

```python
primary_mutation_record_ids = (
    mutation_sample_ids & primary_sample_ids
)

profiled_without_mutation_records = (
    sequenced_primary_ids - primary_mutation_record_ids
)

print(
    "Profiled primary samples with no mutation rows:",
    len(profiled_without_mutation_records)
)

profiled_without_mutation_records
```

```python
profiled_without_selected_mutations = (
    sequenced_primary_ids - protein_altering_samples
)

print(
    "Profiled primary samples with zero selected protein-altering mutations:",
    len(profiled_without_selected_mutations)
)

profiled_without_selected_mutations
```

```python
mutation_matrix_profiled = mutation_matrix.reindex(
    sorted(sequenced_primary_ids),
    fill_value=0
)

print("Rows (profiled primary samples):",
      mutation_matrix_profiled.shape[0])

print("Columns (genes):",
      mutation_matrix_profiled.shape[1])
```

```python
zero_profiled_rows = (
    mutation_matrix_profiled.sum(axis=1) == 0
)

print(
    "Profiled primary samples with zero selected protein-altering mutations:",
    zero_profiled_rows.sum()
)
mutation_matrix_profiled.index[
    zero_profiled_rows
].tolist()
```

```python
from pathlib import Path

mutation_matrix_path = Path(
    "../data/processed/gbm_primary_profiled_mutation_matrix.tsv"
)

mutation_matrix_profiled.to_csv(
    mutation_matrix_path,
    sep="\t",
    index=True
)

print("Saved:", mutation_matrix_path)
print("Exists:", mutation_matrix_path.exists())
```
